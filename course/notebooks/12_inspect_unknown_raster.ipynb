# Inspect Unknown Raster

**Learning objective:** Inspect a local raster before choosing whether it should be styled as continuous or categorical.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene, inspect_raster
from course_helpers import BOUNDS, ensure_output_dir, export_scene, soil_classes, terrain, write_raster

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
unknown = soil_classes()
unknown_path = write_raster(output_dir / '12_unknown_layer.tif', unknown)
profile = inspect_raster(unknown_path)
print(profile)
scene = TerrainScene.from_bbox(*BOUNDS, title='Unknown raster after inspection')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_raster('Inspected unknown layer', unknown_path, style='indiasat_lulc')
html_path, _ = export_scene(scene, '12_inspected_unknown_raster.html')
assert profile['suggested_style'] == 'categorical'
print(f'Exported {html_path}')
```
