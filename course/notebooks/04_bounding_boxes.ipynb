# Bounding Boxes

**Learning objective:** Learn how WGS84 bounds control the latitude/longitude frame of a terrain model.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, export_scene, terrain
```

```python
dem = terrain('plateau')
small_bounds = (76.16, 18.06, 76.31, 18.21)
wide_bounds = BOUNDS
for label, bounds in [('small', small_bounds), ('wide', wide_bounds)]:
    scene = TerrainScene.from_bbox(*bounds, title=f'{label.title()} bounding box')
    scene.add_dem_array(dem, bounds=bounds)
    html_path, _ = export_scene(scene, f'04_{label}_bounds.html')
    print(f'{label} bounds {bounds} -> {html_path}')
```
