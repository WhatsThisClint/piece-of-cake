# Custom DEM Provider

**Learning objective:** Use a small provider class to supply DEM values without hard-coding the array into the scene.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, export_scene, terrain
```

```python
class SyntheticDemProvider:
    def fetch_dem(self, bounds, *, width, height):
        size = min(width, height)
        return terrain('watershed', size=size), bounds

scene = TerrainScene.from_bbox(*BOUNDS, title='Provider-backed synthetic DEM')
scene.add_dem(provider=SyntheticDemProvider(), width=72, height=72)
html_path, _ = export_scene(scene, '05_custom_provider.html')
print(f'Provider exported {html_path}')
```
