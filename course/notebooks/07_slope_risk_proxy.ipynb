# Slope Risk Proxy

**Learning objective:** Derive a simple slope-risk raster from DEM gradients and drape it over terrain.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, ensure_output_dir, export_scene, slope_risk, terrain, write_raster

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
risk = slope_risk(dem)
risk_path = write_raster(output_dir / '07_slope_risk.tif', risk)
scene = TerrainScene.from_bbox(*BOUNDS, title='Slope risk proxy')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_raster('Slope risk', risk_path, style='risk')
html_path, _ = export_scene(scene, '07_slope_risk.html')
print(f'Risk classes: {sorted(set(risk.ravel().tolist()))}')
print(f'Exported {html_path}')
```
