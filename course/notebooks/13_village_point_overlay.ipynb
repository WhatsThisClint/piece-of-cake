# Village Overlay

**Learning objective:** Create synthetic village polygons and show labels plus hover fields over the 3D terrain.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from pathlib import Path

if not Path('course_helpers.py').exists():
    from urllib.request import urlretrieve

    COURSE_HELPERS_URL = 'https://raw.githubusercontent.com/WhatsThisClint/piece-of-cake/main/course/notebooks/course_helpers.py'
    urlretrieve(COURSE_HELPERS_URL, 'course_helpers.py')

from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, ensure_output_dir, export_scene, terrain, write_villages

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
villages_path = write_villages(output_dir / '13_villages.gpkg')
scene = TerrainScene.from_bbox(*BOUNDS, title='Village overlay')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_vector(
    'Villages',
    villages_path,
    layer='villages',
    label_column='village_name',
    hover_fields=['village_name', 'population', 'risk_class'],
    style={'type': 'single', 'outline': '#1d4ed8'},
)
html_path, _ = export_scene(scene, '13_village_overlay.html')
print(f'Village overlay exported to {html_path}')
```
