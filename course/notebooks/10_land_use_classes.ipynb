# Land Use Classes

**Learning objective:** Use categorical land-use values and the built-in IndiaSat LULC style.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, ensure_output_dir, export_scene, lulc, terrain, write_raster

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
land_use = lulc()
lulc_path = write_raster(output_dir / '10_lulc.tif', land_use)
scene = TerrainScene.from_bbox(*BOUNDS, title='Categorical land use classes')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_raster('Land use', lulc_path, style='indiasat_lulc')
html_path, _ = export_scene(scene, '10_land_use_classes.html')
print(f'LULC classes in sample: {sorted(set(land_use.ravel().tolist()))}')
print(f'Exported {html_path}')
```
