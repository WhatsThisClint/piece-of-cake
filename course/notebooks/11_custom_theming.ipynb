# Custom Theming

**Learning objective:** Define a custom categorical style when an organization brings its own class codes.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from pathlib import Path

if not Path('course_helpers.py').exists():
    from urllib.request import urlretrieve

    COURSE_HELPERS_URL = 'https://raw.githubusercontent.com/WhatsThisClint/piece-of-cake/main/course/notebooks/course_helpers.py'
    urlretrieve(COURSE_HELPERS_URL, 'course_helpers.py')

from piece_of_cake import StyleProfile, TerrainScene
from course_helpers import BOUNDS, ensure_output_dir, export_scene, soil_classes, terrain, write_raster

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
soil = soil_classes()
soil_path = write_raster(output_dir / '11_soil_classes.tif', soil)
soil_style = StyleProfile(
    kind='categorical',
    opacity=0.72,
    classes={
        1: {'label': 'Shallow soil', 'color': '#fef3c7'},
        2: {'label': 'Medium soil', 'color': '#d97706'},
        3: {'label': 'Deep black soil', 'color': '#374151'},
        4: {'label': 'Rocky patch', 'color': '#9ca3af'},
    },
)
scene = TerrainScene.from_bbox(*BOUNDS, title='Custom soil theme')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_raster('Soil classes', soil_path, style=soil_style)
html_path, _ = export_scene(scene, '11_custom_theming.html')
print(f'Custom theme exported to {html_path}')
```
