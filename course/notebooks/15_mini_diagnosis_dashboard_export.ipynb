# Mini Diagnosis Dashboard Export

**Learning objective:** Combine relief, risk, canopy, land use, and village attributes into one HTML diagnosis model.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene
from course_helpers import (
    BOUNDS,
    ensure_output_dir,
    export_scene,
    lulc,
    slope_risk,
    terrain,
    tree_canopy,
    write_raster,
    write_villages,
)

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
risk_path = write_raster(output_dir / '15_risk.tif', slope_risk(dem))
canopy_path = write_raster(output_dir / '15_canopy.tif', tree_canopy())
lulc_path = write_raster(output_dir / '15_lulc.tif', lulc())
villages_path = write_villages(output_dir / '15_villages.gpkg')
scene = TerrainScene.from_bbox(*BOUNDS, title='Mini diagnosis dashboard terrain')
scene.vertical_exaggeration = 1.2
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_raster('Slope risk', risk_path, style='risk')
scene.add_raster('Tree canopy', canopy_path, style='tree_canopy')
scene.add_raster('Land use', lulc_path, style='indiasat_lulc')
scene.add_vector(
    'Villages',
    villages_path,
    layer='villages',
    label_column='village_name',
    hover_fields=['village_name', 'population', 'risk_class', 'sc_st_pct'],
    style={'type': 'single', 'outline': '#111827'},
)
html_path, html = export_scene(scene, '15_mini_diagnosis_dashboard.html')
assert 'Slope risk' in html and 'Tree canopy' in html and 'Land use' in html
print(f'Final dashboard exported to {html_path}')
```
