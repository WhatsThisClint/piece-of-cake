# Watershed Relief Story

**Learning objective:** Turn relief into a simple watershed interpretation: uplands, valley floor, and possible flow concentration.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, export_scene, terrain
```

```python
dem = terrain('watershed')
low_threshold = float(dem.mean() - dem.std() * 0.45)
high_threshold = float(dem.mean() + dem.std() * 0.65)
print(f'Valley-floor threshold: below {low_threshold:.1f} m')
print(f'Upland threshold: above {high_threshold:.1f} m')
scene = TerrainScene.from_bbox(*BOUNDS, title='Watershed relief interpretation')
scene.vertical_exaggeration = 1.25
scene.add_dem_array(dem, bounds=BOUNDS)
html_path, _ = export_scene(scene, '06_watershed_relief.html')
print(f'Relief story exported to {html_path}')
```
