# Tree Canopy Layer

**Learning objective:** Create a synthetic tree-canopy raster and style it with the built-in tree canopy profile.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from pathlib import Path

if not Path('course_helpers.py').exists():
    from urllib.request import urlretrieve

    COURSE_HELPERS_URL = 'https://raw.githubusercontent.com/WhatsThisClint/piece-of-cake/main/course/notebooks/course_helpers.py'
    urlretrieve(COURSE_HELPERS_URL, 'course_helpers.py')

from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, ensure_output_dir, export_scene, terrain, tree_canopy, write_raster

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
canopy = tree_canopy()
canopy_path = write_raster(output_dir / '08_tree_canopy.tif', canopy)
scene = TerrainScene.from_bbox(*BOUNDS, title='Tree canopy on terrain')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_raster('Tree canopy', canopy_path, style='tree_canopy')
html_path, _ = export_scene(scene, '08_tree_canopy.html')
print(f'Canopy percent range: {canopy.min():.1f} to {canopy.max():.1f}')
print(f'Exported {html_path}')
```
