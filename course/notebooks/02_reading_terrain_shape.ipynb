# Reading Terrain Shape

**Learning objective:** Compare ridge, valley, and plateau terrain patterns and see how exaggeration changes interpretation.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from pathlib import Path

if not Path('course_helpers.py').exists():
    from urllib.request import urlretrieve

    COURSE_HELPERS_URL = 'https://raw.githubusercontent.com/WhatsThisClint/piece-of-cake/main/course/notebooks/course_helpers.py'
    urlretrieve(COURSE_HELPERS_URL, 'course_helpers.py')

from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, export_scene, terrain
```

```python
for kind, exaggeration in [('ridge', 1.0), ('valley', 1.35), ('plateau', 0.9)]:
    scene = TerrainScene.from_bbox(*BOUNDS, title=f'{kind.title()} terrain')
    scene.vertical_exaggeration = exaggeration
    scene.add_dem_array(terrain(kind, size=72), bounds=BOUNDS)
    html_path, _ = export_scene(scene, f'02_{kind}_terrain.html')
    print(f'{kind}: {html_path}')
```
