# NDVI Layer

**Learning objective:** Create a synthetic NDVI layer and drape it as a continuous vegetation surface.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene
from course_helpers import BOUNDS, ensure_output_dir, export_scene, ndvi, terrain, write_raster

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
ndvi_values = ndvi()
ndvi_path = write_raster(output_dir / '09_ndvi.tif', ndvi_values)
scene = TerrainScene.from_bbox(*BOUNDS, title='NDVI vegetation signal')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_raster('NDVI', ndvi_path, style='ndvi')
html_path, _ = export_scene(scene, '09_ndvi.html')
print(f'NDVI range: {ndvi_values.min():.2f} to {ndvi_values.max():.2f}')
print(f'Exported {html_path}')
```
