# Choosing Vector Columns

**Learning objective:** Inspect vector attributes and choose which columns belong in labels and hover text.


This notebook uses synthetic, deterministic data generated on your machine. It does not download anything, and all exported files go into `course_outputs/`.


```python
from piece_of_cake import TerrainScene, inspect_vector
from course_helpers import BOUNDS, ensure_output_dir, export_scene, terrain, write_village_scores

output_dir = ensure_output_dir()
```

```python
dem = terrain('watershed')
scores_path = write_village_scores(output_dir / '14_village_scores.gpkg')
summary = inspect_vector(scores_path, layer='village_scores')
print(summary['suggested_columns'])
chosen_label = 'village_code'
chosen_hover = ['village_code', 'population', 'sc_st_pct', 'risk_score']
scene = TerrainScene.from_bbox(*BOUNDS, title='Chosen vector columns')
scene.add_dem_array(dem, bounds=BOUNDS)
scene.add_vector(
    'Village score attributes',
    scores_path,
    layer='village_scores',
    label_column=chosen_label,
    hover_fields=chosen_hover,
    style={'type': 'single', 'outline': '#7c3aed'},
)
html_path, _ = export_scene(scene, '14_chosen_vector_columns.html')
column_names = [column['name'] for column in summary['columns']]
assert chosen_label in column_names
print(f'Exported {html_path}')
```
